```python
import time
import numpy as np
```

```python
def compare_addition_speed():
    py_list = list(range(10000))
    np_array = np.arange(10000)

    start = time.time()
    list_result = [x + 5 for x in py_list]
    list_time = time.time() - start

    start = time.time()
    array_result = np_array + 5
    array_time = time.time() - start

    print("Python List Time:", list_time, "seconds")
    print("NumPy Array Time:", array_time, "seconds")
```

```python
compare_addition_speed()
```
