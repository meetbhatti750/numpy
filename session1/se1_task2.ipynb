```python
import numpy as np
import sys
```

```python
py_list = list(range(1, 1001))
np_array = np.arange(1, 1001)
```

```python

list_memory = sys.getsizeof(py_list)
array_memory = np_array.nbytes
```

```python

print("Python List Memory:", list_memory, "bytes")
```

```python


print("NumPy Array Memory:", array_memory, "bytes")
```
