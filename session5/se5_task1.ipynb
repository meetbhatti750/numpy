```python

import numpy as np
```

```python
my_likes = np.array([120, 150, 200, 180, 250, 300, 350])
friend_likes = np.array([100, 130, 180, 160, 200, 280, 300])
```

```python
combined = np.add(my_likes, friend_likes)
difference = np.subtract(my_likes, friend_likes)
```

```python
print("Combined Likes:", combined)
```

```python
print("Difference:", difference)
```
