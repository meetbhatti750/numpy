```python
import numpy as np
```

```python
steps = np.array([4500, 6200, 7800, 5100, 9000,
                    10500, 6800, 7200, 8400, 5600,
                    12000, 9500, 6300, 7500, 8800,
                    11000, 4900, 8100, 9200, 7000,
                    13500, 10200, 5800, 7600, 8900,
                    11500, 6500, 9800, 12500, 8000  
])
```

```python
average_steps = np.mean(steps)
```

```python
print("Daily Steps:", steps)
```

```python
median_steps = np.median(steps)
```

```python
print("Average Steps:", average_steps)
```

```python
std_steps = np.std(steps)
```

```python
print("Standard Deviation:", std_steps)
```

```python
max_steps = np.max(steps)
```

```python
print("Maximum Steps:", max_steps)
```
