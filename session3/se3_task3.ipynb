```python
import numpy as np
```

```python
prices = np.array([120, 80, 150, 200, 90, 110,
                   75, 180, 250, 60, 130, 100])
```

```python
print("Original:")
print(prices)
```

```python
print("Ravel:")
print(prices.ravel())
```

```python
print("Flatten:")
print(prices.flatten())
```

```python
prices.resize(3, 4)
```

```python
print("Resize:")
print(prices)
```
