```python
import numpy as np
```

```python
spotify = np.array([
    [10, 20, 30],
    [15, 25, 35],
    [12, 22, 32]
])
```

```python
print("Original array:")
print(spotify)
```

```python
print("Using T:")
print(spotify.T)
```

```python
print("Using np.transpose():")
print(np.transpose(spotify))
```
