```python
import numpy as np
```

```python
steps = np.array([
    [3000, 5000],
    [4000, 6000],
    [3500, 5500],
    [4500, 7000],
    [5000, 8000]
])
```

```python
print("Original 2D array:")
print(steps)
```

```python
one_d = steps.reshape(10)
print("1D array:")
print(one_d)
```

```python
two_d = one_d.reshape(5, 2)
print("Back to 2D:")
print(two_d)
```
